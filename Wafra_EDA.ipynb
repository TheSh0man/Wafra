```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("../myenv/egypt_spain_Final_Editon.csv")     
print(df.shape)
df
```

```python
n = df["Country"].nunique()
EFF = 0.50      

df["sw_twh"]    = df["solar_gen_twh"] + df["wind_gen_twh"]              # solar + wind
df["sw_share"]  = 100 * df["sw_twh"] / df["generation_twh"]             # % of sw generation
df["gas_share"] = 100 * df["gas_elec_twh"] / df["generation_twh"]
df["other_twh"] = df["generation_twh"] - df["fossil_elec_twh"] - df["sw_twh"]   # hydro, nuclear, etc.
df
```

## Quality check 

```python
print("Duplicate country-year rows:", df.duplicated(["Country", "Year"]).sum())
print(df.groupby("Country").size())
print("\nColumns with missing values:\n", df.isna().sum()[lambda s: s > 0])

# Each of these should print 0
print("\nsolar+wind differs from renewable_gen_twh:", ((df.sw_twh - df.renewable_gen_twh).abs() > 0.011).sum())
print("gas+oil+coal differs from fossil_elec_twh:",
      ((df.gas_elec_twh + df.oil_elec_twh + df.coal_elec_twh - df.fossil_elec_twh).abs() > 0.011).sum())
print("Shares outside 0-100:", ((df.sw_share < 0) | (df.sw_share > 100)).sum())
print("Negative 'other' generation:", (df.other_twh < 0).sum())

# Does the 50% efficiency make sense? Gas needed for power must be below total gas use (<100%)
print("\nMax implied gas-for-power as % of gas consumption:")
print((100 * df.gas_elec_twh / EFF / df.gas_consumption_twh).groupby(df.Country).max().round(0))
```

 ## How much did solar and wind grow?

```python
fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
sns.lineplot(data=df, x="Year", y="sw_twh", hue="Country", marker="o", ax=ax[0])
sns.lineplot(data=df, x="Year", y="sw_share", hue="Country", marker="o", ax=ax[1])
ax[0].set_ylabel("Solar + wind generation (TWh)")
ax[1].set_ylabel("Share of total generation (%)")
plt.show()
```

## What does the electricity mix look like?

```python
parts = {"gas_elec_twh": "Gas", "oil_elec_twh": "Oil", "coal_elec_twh": "Coal","wind_gen_twh": "Wind", "solar_gen_twh": "Solar", "other_twh": "Other (hydro, nuclear, etc.)"}
fig, axes = plt.subplots(1, n, figsize=(6.5 * n, 4.5))
for ax, (country, g) in zip(axes, df.groupby("Country")):
    mix = g.set_index("Year")[list(parts)].rename(columns=parts)
    (mix.div(mix.sum(axis=1), axis=0) * 100).plot.area(ax=ax, linewidth=0)
    ax.set_title(country); ax.set_ylabel("% of generation")
plt.show()
```

## Did solar and wind replace gas, or just meet new demand?

```python
cols = ["gas_elec_twh", "oil_elec_twh", "coal_elec_twh", "sw_twh", "other_twh", "generation_twh"]
names = ["Gas", "Oil", "Coal", "Solar + wind", "Other", "Total generation"]

change = df.groupby("Country")[cols].last() - df.groupby("Country")[cols].first()   # last year minus first year
change.columns = names

change.T.plot.bar(figsize=(10, 5))
plt.axhline(0, color="black", lw=0.8)
plt.ylabel("Change in generation, first to last year (TWh)")
plt.xticks(rotation=0); plt.show()
print(change.round(1).T)
```

##  How do Egypt and Spain compare?

```python
latest = df[df.Year == df.Year.max()].set_index("Country")
table = latest[["sw_twh", "sw_share", "gas_share", "solar_per_capita_kwh", "wind_per_capita_kwh"]].round(1)
table.columns = ["Solar+wind (TWh)", "Solar+wind share (%)", "Gas share (%)",
                 "Solar kWh/person", "Wind kWh/person"]
print(table.T)

fig, ax = plt.subplots(1, 3, figsize=(14, 4))
for a, col, title in zip(ax, ["sw_twh", "sw_share", "solar_per_capita_kwh"],
                         ["Solar + wind (TWh)", "Solar + wind share (%)", "Solar per person (kWh)"]):
    latest[col].plot.bar(ax=a, color=sns.color_palette("deep", n)); a.set_title(title); a.set_xlabel("")
    a.tick_params(axis="x", rotation=0)
plt.tight_layout(); plt.show()
```

## How much gas could solar and wind have avoided?

```python
df["avoided_gas_twh"]  = df.sw_twh / EFF       # gas energy needed to make the same electricity
df["avoided_low"]      = df.sw_twh / 0.60      # if gas plants were more efficient (less gas needed)
df["avoided_high"]     = df.sw_twh / 0.35      # if gas plants were less efficient (more gas needed)
df["avoided_mmbtu"]    = df.avoided_gas_twh * 3_412_141
df["displacement_pct"] = 100 * df.sw_twh / (df.sw_twh + df.gas_elec_twh)   # does not depend on efficiency

fig, axes = plt.subplots(1, n, figsize=(6.5 * n, 4.5))
for ax, (country, g) in zip(axes, df.groupby("Country")):
    ax.fill_between(g.Year, g.avoided_low, g.avoided_high, alpha=0.25, color="green", label="35-60% efficiency range")
    ax.plot(g.Year, g.avoided_gas_twh, "o-", color="darkgreen", label="Base case (50%)")
    ax.set_title(country); ax.set_ylabel("Avoided gas (TWh of gas energy)"); ax.legend()
plt.show()

print(df.groupby("Country")[["avoided_low", "avoided_gas_twh", "avoided_high"]].sum().round(0))
print(df.pivot(index="Year", columns="Country", values="displacement_pct").round(1))
```

## Which renewable source contributes more: Solar or Wind

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, country in zip(axes, df["Country"].dropna().unique()):

    data = df[df["Country"] == country].sort_values("Year")

    ax.stackplot(
        data["Year"],
        data["solar_gen_twh"],
        data["wind_gen_twh"],
        labels=["Solar", "Wind"],
        alpha=0.8
    )

    ax.set_title(f"{country}: Solar vs Wind")
    ax.set_xlabel("Year")
    ax.set_ylabel("Generation (TWh)")
    ax.legend()

plt.tight_layout()
plt.show()
```

##  Is gas-fired electricity generation decreasing as renewables increase?

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, (country, group) in zip(
    axes, df.groupby("Country")
):

    group = group.sort_values("Year")

    ax.plot(
        group["Year"],
        group["renewable_gen_twh"],
        marker="o",
        label="Solar + Wind"
    )

    ax.plot(
        group["Year"],
        group["gas_elec_twh"],
        marker="o",
        label="Gas Generation"
    )

    ax.set_title(country)
    ax.set_xlabel("Year")
    ax.set_ylabel("Generation (TWh)")
    ax.legend()

plt.tight_layout()
plt.show()
```

## How has gas dependency in electricity generation changed?

```python
fig, ax = plt.subplots(figsize=(10, 5))

for country, group in df.groupby("Country"):

    group = group.sort_values("Year")

    gas_share = (
        group["gas_elec_twh"] /
        group["generation_twh"] * 100
    )

    ax.plot(
        group["Year"],
        gas_share,
        marker="o",
        label=country
    )

ax.set_title("Gas Share of Total Electricity Generation")
ax.set_xlabel("Year")
ax.set_ylabel("Gas Share (%)")
ax.legend()

plt.show()
```

## Does electricity demand grow faster than renewable generation?

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, (country, group) in zip(
    axes, df.groupby("Country")
):

    group = group.sort_values("Year")

    ax.plot(
        group["Year"],
        group["demand_twh"],
        marker="o",
        label="Electricity Demand"
    )

    ax.plot(
        group["Year"],
        group["renewable_gen_twh"],
        marker="o",
        label="Solar + Wind"
    )

    ax.set_title(country)
    ax.set_xlabel("Year")
    ax.set_ylabel("TWh")
    ax.legend()

plt.tight_layout()
plt.show()
```

## How do Egypt and Spain compare in solar and wind generation per capital?

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for country, group in df.groupby("Country"):

    group = group.sort_values("Year")

    axes[0].plot(
        group["Year"],
        group["solar_per_capita_kwh"],
        marker="o",
        label=country
    )

    axes[1].plot(
        group["Year"],
        group["wind_per_capita_kwh"],
        marker="o",
        label=country
    )

axes[0].set_title("Solar Generation per Capita")
axes[0].set_ylabel("kWh per Person")
axes[0].set_xlabel("Year")
axes[0].legend()

axes[1].set_title("Wind Generation per Capita")
axes[1].set_ylabel("kWh per Person")
axes[1].set_xlabel("Year")
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Does natural gas consumption follow the same trend as gas electricity generation?

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, (country, group) in zip(
    axes, df.groupby("Country")
):

    group = group.sort_values("Year")

    ax.plot(
        group["Year"],
        group["gas_consumption_twh"],
        marker="o",
        label="Total Gas Consumption"
    )

    ax.plot(
        group["Year"],
        group["gas_elec_twh"],
        marker="o",
        label="Gas Electricity Generation"
    )

    ax.set_title(country)
    ax.set_xlabel("Year")
    ax.set_ylabel("TWh")
    ax.legend()

plt.tight_layout()
plt.show()
```

## Estimated Avoided Gas

```python
# Assumed gas power plant efficiency
efficiency = 0.50

# Estimate avoided gas input
df["avoided_gas_twh"] = (
    df["renewable_gen_twh"] / efficiency
)

# Plot yearly avoided gas
fig, ax = plt.subplots(figsize=(11, 5))

sns.barplot(
    data=df,
    x="Year",
    y="avoided_gas_twh",
    hue="Country",
    ax=ax
)

ax.set_title("Estimated Avoided Gas Input")
ax.set_ylabel("Equivalent Gas Input (TWh)")
ax.set_xlabel("Year")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
