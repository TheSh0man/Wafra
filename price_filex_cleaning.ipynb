```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
# Read the Excel file
file_path = r"E:\DEPI project DATA\prices\CMO-Historical-Data-Monthly.xlsx"

df = pd.read_excel(
    file_path,
    sheet_name="Monthly Prices",
    header=None,
    skiprows=6,
    usecols=[0, 7, 8, 9]
)
```

```python
# Rename columns
df.columns = [
    "Date",
    "Natural_Gas_US",
    "Natural_Gas_Europe",
    "LNG_Japan"
]
```

```python
# Remove rows without a valid date
df = df[df["Date"].astype(str).str.match(r"^\d{4}M\d{2}$")]
```

```python
# Keep data from 2015 onwards
df = df[df["Date"].astype(str).str[:4].astype(int) >= 2015]
```

```python
# Replace missing value symbols
df = df.replace(["…", "..."], pd.NA)
```

```python
# Convert prices to numeric
price_columns = [
    "Natural_Gas_US",
    "Natural_Gas_Europe",
    "LNG_Japan"
]

for col in price_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Check duplicates
print("Duplicates:", df.duplicated(subset=["Date"]).sum())
```

```python
# Sort by date
df = df.sort_values("Date")
```

```python
# Save cleaned data
df.to_csv(
    r"E:\DEPI project DATA\prices\Gas_Prices_Monthly_Cleaned.csv",
    index=False
)

print(df.head())
print(df.info())
```

```python
# Read the Excel file
file_path = r"E:\DEPI project DATA\prices\CMO-Historical-Data-Annual.xlsx"
df = pd.read_excel(
    file_path,
    sheet_name="Annual Prices (Nominal)",
    header=None,
    skiprows=8,
    usecols=[0, 7, 8, 9]
)
```

```python
# Rename columns
df.columns = [
    "Year",
    "Natural_Gas_US",
    "Natural_Gas_Europe",
    "LNG_Japan"
]
```

```python
# Keep valid years only
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df = df.dropna(subset=["Year"])
df["Year"] = df["Year"].astype(int)
```

```python

# Keep data from 2015 onwards
df = df[df["Year"] >= 2015]
```

```python
# Replace missing value symbols
df = df.replace(["…", "..."], pd.NA)
```

```python
# Convert prices to numeric
price_columns = [
    "Natural_Gas_US",
    "Natural_Gas_Europe",
    "LNG_Japan"
]

for col in price_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Check duplicates
print("Duplicates:", df.duplicated(subset=["Year"]).sum())
```

```python
# Sort by year
df = df.sort_values("Year")
```

```python
# Save cleaned data
df.to_csv(
    r"E:\DEPI project DATA\prices\Gas_Prices_Annual_Cleaned.csv",
    index=False
)
```
